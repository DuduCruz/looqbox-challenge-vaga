### **1st Task**

Primeiramente vamos verificar quais são os produtos de maior valor na empresa.

---
First of all, we'll check the 10 most expensive products in the company.

```python
query_produtos_caros = text("""
    SELECT
        PRODUCT_COD,
        PRODUCT_NAME,
        PRODUCT_VAL
    FROM data_product
    ORDER BY PRODUCT_VAL DESC
    LIMIT 10;
""")
```

### **2nd Task**

Agora vamos descobrir em quais seções os departamentos de "BEBIDAS" e "PADARIA" se encontram.

---
Now we'll find out in which sections the "BEVERAGES" and "BAKERY" departaments are.

```python
query_secoes_departamento = text("""
    SELECT DISTINCT
        DEP_NAME,
        SECTION_NAME
    FROM data_product
    WHERE DEP_NAME IN ('BEBIDAS', 'PADARIA')
    ;
""")
```

### **3rd Task**

Por último vamos buscar saber o total de vendas de produtos (em $) de cada área de negócio no primeiro trimestre de 2019.

---
Lastly, we will determine the total product sales (in $) for each business area in the first quarter of 2019.

```python
query_vendas_trimestre = text("""
    SELECT
        cadastro.BUSINESS_NAME AS AREA_DE_NEGOCIO,
        SUM(vendas.SALES_VALUE) AS TOTAL_VENDIDO
    FROM data_store_cad AS cadastro
    INNER JOIN data_store_sales AS vendas
        ON cadastro.STORE_CODE = vendas.STORE_CODE
    WHERE vendas.DATE BETWEEN '2019-01-01' AND '2019-03-31'
    GROUP BY cadastro.BUSINESS_NAME;
""")
```

# **CASE 1 - FUNÇÃO DINÂMICA DE EXTRAÇÃO DE DADOS | DYNAMIC DATA EXTRACTION**

```python
import pandas as pd
from sqlalchemy import text

# FUNÇÃO DINÂMICA DE EXTRAÇÃO DE DADOS
def retrieve_data(product_code, store_code, date, engine_conexao):

    # Extraindo a data de início e a data de fim de dentro da lista passada pelo usuário
    data_inicio = date[0]
    data_fim = date[1]

    # Construção da query SQL utilizando f-strings para injetar os parâmetros dinamicamente.

    query_dinamica = text(f"""
        SELECT *
        FROM data_product_sales
        WHERE PRODUCT_CODE = {product_code}
          AND STORE_CODE = {store_code}
          AND DATE BETWEEN '{data_inicio}' AND '{data_fim}';
    """)

    print("--- Query Gerada com Sucesso ---")
    print(query_dinamica)
    print("-" * 32)

    try:
        # Abre a conexão com o banco de dados utilizando o gerenciador de contexto (with)
        with engine_conexao.connect() as conexao:

            df_resultado = pd.read_sql(query_dinamica, con=conexao)

            # Avisa o usuário caso a consulta funcione, mas não encontre nenhum registro no banco
            if df_resultado.empty:
                print("Atenção: A consulta funcionou, mas nenhum dado foi encontrado para esses filtros.")
            else:
                print(f" Sucesso! {len(df_resultado)} linhas retornadas pelo banco.")

            return df_resultado

    except Exception as erro:
        # Se houver erro de sintaxe ou coluna inexistente, exibe uma mensagem clara
        print(f"❌ Erro ao executar a consulta dinâmica.")
        print(f"Detalhes do erro: {erro}")
        return None


# TESTE DE VALIDAÇÃO
# Quando for rodar, certifique-se de usar códigos que de fato existam no banco!

print("Executando teste da função dinâmica...\n")

codigo_produto = 18                     #Preencha com dados reais do banco de dados
codigo_loja = 1                         #Preencha com dados reais do banco de dados
periodo = ['2019-01-01', '2019-01-31']  #Preencha com dados reais do banco de dados

# Chamando a nossa função e guardando o resultado na variável 'my_data'

my_data = retrieve_data(
    product_code=codigo_produto,
    store_code=codigo_loja,
    date=periodo,
    engine_conexao=engine
)

if my_data is not None:
    display(my_data.head())
```

```python
query_1_cliente = text("""
SELECT
      STORE_CODE,
      STORE_NAME,
      START_DATE,
      END_DATE,
      BUSINESS_NAME,
      BUSINESS_CODE
FROM data_store_cad
""")
```

```python
query_2_cliente = text("""
SELECT
        STORE_CODE,
        DATE,
        SALES_VALUE,
        SALES_QTY
FROM data_store_sales
WHERE DATE BETWEEN '2019-01-01' AND '2019-12-31'
""")
```

```python
import pandas as pd
from sqlalchemy import text


def gerar_matriz_ticket_medio(engine_conexao):

    # Executar as queries do cliente, filtra o período solicitado, calcula o Ticket Médio (TM) e renderiza a matriz Loja vs Categoria.

    print("Iniciando o processamento dos dados do cliente...")

    # Buscando os dados do banco de dados de forma isolada

    with engine_conexao.connect() as conexao:
        df_lojas = pd.read_sql(query_1_cliente, con=conexao)
        df_vendas = pd.read_sql(query_2_cliente, con=conexao)

    print(f"-> Dados carregados: {len(df_lojas)} lojas e {len(df_vendas)} registros de vendas.")

    df_vendas['DATE'] = pd.to_datetime(df_vendas['DATE'])

    # Aplicar o filtro de data exigido pelo cliente (4º Trimestre de 2019)

    data_inicio = '2019-10-01' #Caso queira ajustar o período desejado - ajustar nesse campo
    data_fim = '2019-12-31'    #Caso queira ajustar o período desejado - ajustar nesse campo
    df_vendas_filtrado = df_vendas[df_vendas['DATE'].between(data_inicio, data_fim)].copy()

    print(f"-> Vendas após filtragem do período ({data_inicio} até {data_fim}): {len(df_vendas_filtrado)} registros.")

    # Cruzar as tabelas utilizando a coluna comum STORE_CODE

    df_consolidado = pd.merge(df_vendas_filtrado, df_lojas, on='STORE_CODE', how='inner')

    # Agrupar por Loja e Categoria para somar os valores financeiros e quantidades
    df_agrupado = df_consolidado.groupby(['STORE_NAME', 'BUSINESS_NAME']).agg({
        'SALES_VALUE': 'sum',
        'SALES_QTY': 'sum'
    }).reset_index()

    # Calcular a métrica de Ticket Médio (TM = Valor Total / Quantidade Total)

    df_agrupado['TM'] = (df_agrupado['SALES_VALUE'] / df_agrupado['SALES_QTY']).round(2) #Usei o round(2) para deixar formatado, com o valor dos centavos


    tabela_tm = df_agrupado[['STORE_NAME', 'BUSINESS_NAME', 'TM']].copy() # Filtrar apenas as colunas solicitadas para o relatório
    tabela_tm.columns = ['Loja', 'Categoria', 'TicketMédio'] # Renomeamos as colunas para o português padrão exigido
    tabela_tm = tabela_tm.sort_values(by='Loja').reset_index(drop=True) # Ordenar alfabeticamente pelo nome da Loja para melhor visualização


    tabela_tm.index = tabela_tm.index + 1

    print("\n Tabela de Ticket Médio gerada com sucesso!")
    print("-" * 60)

    display(tabela_tm)

    return tabela_tm


df_resultado_caso2 = gerar_matriz_ticket_medio(engine)
```

```python
import pandas as pd
import plotly.express as px
from sqlalchemy import text

query_imdb_oficial = text("""
    SELECT
        Title,
        Rating
    FROM IMDB_movies
    WHERE Rating IS NOT NULL
    ORDER BY Rating DESC
    LIMIT 10;
""")

def gerar_grafico_oficial_imdb(engine_conexao):

    print("Conectando ao banco e extraindo dados oficiais do IMDB...")

    try:
        with engine_conexao.connect() as conexao:
            df_movies = pd.read_sql(query_imdb_oficial, con=conexao)

        print("-> Dados carregados com sucesso! Construindo o gráfico interativo...")


        fig = px.bar(
            df_movies,
            x='Rating',
            y='Title',
            orientation='h',
            title='<b>Top 10 Filmes Mais Bem Avaliados no IMDB</b>',
            labels={'Rating': 'Nota de Avaliação (Rating)', 'Title': 'Título do Filme'},
            text='Rating',
            color='Rating',
            color_continuous_scale='Blues'
        )

        fig.update_layout(
            title_font_size=18,
            xaxis=dict(range=[0, 10]),
            yaxis={'categoryorder': 'total ascending'},
            showlegend=False,
            coloraxis_showscale=False,
            height=500
        )

        fig.update_traces(
            textposition='outside',
            textfont=dict(size=12, color='black'))

        fig.show(renderer="colab")

    except Exception as erro:
        print(f"Ocorreu um erro ao processar os dados oficiais.")
        print(f"Detalhes do erro para análise do Dev: {erro}")

gerar_grafico_oficial_imdb(engine)
```

A visualização conforme a coluna "rating" nos traz diversos filmes com as mesmas notas, uma solução seria elencar conforme quantidade de votos, ou até mesmo utilizar a coluna "metascore".
